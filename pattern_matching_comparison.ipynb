# Einleitung

Pattern-Matching-Algorithmen sind Algorithmen zum Musterabgleich und erstellen ein wichtiges Werkzeug in der Informatik dar, wenn es darum geht, bestimmte Muster in größeren Datenmengen zu finden (1). # Brute-Force

<b>1. Definition und theoretische Grundlagen <br></b>
<br>Der Brute-Force-Algorithmus (auf Deutsch auch als "Methode der rohen Gewalt" oder "erschöpfende Suche" bezeichnet) stellt den einfachsten und intuitivsten Ansatz zur Lösung des Pattern-Matching-Problems dar. Bei diesem Problem geht es darum, ein bestimmtes Muster (Pattern) innerhalb eines längeren Textes zu finden (4).
Die Grundidee des Brute-Force-Algorithmus besteht darin, systematisch alle möglichen Lösungskandidaten zu überprüfen, bis entweder eine Lösung gefunden wird oder alle Möglichkeiten ausgeschöpft sind (9). Im Kontext der Mustererkennung bedeutet dies, dass das gesuchte Muster mit jedem möglichen Teilstring des Textes verglichen wird. <br>
<b><br>Zeitkomplexität:<br></b>
•	Worst-Case: O(n·m), wobei n die Länge des Textes und m die Länge des Musters ist (4) <br>
•	Best-Case: O(n) tritt ein, wenn das erste Zeichen des Musters in keinem der geprüften Textabschnitte vorkommt (9) <br>
•	Average-Case: Nähert sich in vielen realen Anwendungen O(n+m) <br>
<b><br> Raumkomplexität: <br> </b>
•	O(1), da nur konstanter zusätzlicher Speicherplatz benötigt wird (9) <br>
Diese Komplexitäten machen den Brute-Force-Algorithmus für sehr große Texte oder Muster potenziell ineffizient, was zur Entwicklung fortschrittlicherer Algorithmen wie KMP und Boyer-Moore geführt hat (5). <br>

<b><br>2. Funktionsweise des Brute-Force-Algorithmus<br></b>
<br> Der Brute-Force-Algorithmus zur Mustererkennung funktioniert nach folgendem Prinzip: <br>
1.	Positionierung des Suchmusters am Anfang des Textes<br>
2.	Zeichenweiser Vergleich des Musters mit dem entsprechenden Textabschnitt<br>
3.	Bei Nichtübereinstimmung: Verschiebung des Musters um eine Position nach rechts<br>
4.	Wiederholung der Schritte 2-3, bis entweder ein vollständiges Match gefunden wurde oder das Ende des Textes erreicht ist (9)<br>


```python
def brute_force_string_match(text, pattern):
    n = len(text)
    m = len(pattern)
    for i in range(n-m+1):
        j = 0
        while j < m and text[i+j] == pattern[j]:
            j+=1
            if j == m:
                return i #found the pattern
    return -1 # Not found
```

Dieser Algorithmus durchsucht den Text von links nach rechts und vergleicht an jeder Position, ob das Muster übereinstimmt. Das Verschieben um immer genau eine Position ist charakterisitisch für den Brute-Force-Ansatz und unterscheidet ihn von optimierten Verfahren (7). <br>

<b><br> 3. Detaillierte Analyse der Zeitkomplexität <br></b>

Die Zeitkomplexität des Brute-Force-Algorithmus lässt sich anhand der Anzahl der durchgeführten Zeichenvergleiche bestimmen: <br>
<b><br>Worst-Case-Szenario: O(n·m) <br></b>
Ein Worst-Case-Szenario tritt ein, wenn das Muster fast mit dem Text übereinstimmt, aber im letzten Zeichen eine Abweichung auftritt. Beispielsweise: <br>
<br>•	Text: "AAAAAAAAAAAAAAAB" <br>
•	Muster: "AAAAAB" <br>
<br>Hier wird der Algorithmus für jeden Startindex i = 0, 1, 2, ..., n-m fast m Vergleiche durchführen, bevor er feststellt, dass keine Übereinstimmung vorliegt. Das ergibt insgesamt bis zu (n-m+1)·m Vergleiche, was O(n·m) entspricht (4).<br>
<b><br>Best-Case-Szenario: O(n)<br></b>
Der günstigste Fall tritt ein, wenn das erste Zeichen des Musters in keinem der zu überprüfenden Teilstrings vorkommt. In diesem Fall wird nur ein Vergleich pro Startposition benötigt, was zu n-m+1 Vergleichen führt, also O(n) (9). <br>
<b><br>Average-Case-Analyse<br></b>
In der Praxis liegt die durchschnittliche Anzahl an Vergleichen oft deutlich unter dem Worst-Case. Bei zufällig verteilten Texten beträgt die erwartete Anzahl der Vergleiche etwa 2(n-m+1), was ebenfalls O(n) entspricht, wenn m deutlich kleiner ist als n (3).<br>

<b><br>4. Vergleich mit KMP und Boyer-Moore <br></b>
<br>Der Brute-Force-Algorithmus unterscheidet sich in mehreren wesentlichen Punkten von den fortschrittlicheren Musterkennungsalgorithmen KMP (Knuth-Morris-Pratt) und Boyer-Moore: <br>
Brute-Force vs. KMP-Algorithmus <br>
<b><br>Gemeinsamkeiten:<br></b>
•	Beide Algorithmen scannen den Text von links nach rechts<br>
•	Beide vergleichen das Muster zeichenweise mit dem Text<br>
<b><br>Unterschiede:<br></b>
•	KMP nutzt ein Preprocessing des Musters, um redundante Vergleiche zu vermeiden<br>
•	KMP erreicht eine Zeitkomplexität von O(n+m) statt O(n·m) (4)<br>
•	Bei einem Mismatch kann KMP das Muster um mehr als eine Position verschieben<br>
•	Im Beispiel von "ABABAABABAB" mit Muster "ABABAB" benötigt Brute-Force 20 Vergleiche, während KMP nur 14 Vergleiche durchführt (4)<br>
<b><br>Brute-Force vs. Boyer-Moore-Algorithmus<br></b>
<b><br>Gemeinsamkeiten:<br></b>
•	Beide Algorithmen suchen nach exakten Übereinstimmungen<br>
•	Beide sind für beliebige Textarten anwendbar<br>
<b><br>Unterschiede:<br></b>
•	Boyer-Moore scannt das Muster von rechts nach links, nicht von links nach rechts<br>
•	Boyer-Moore nutzt zwei Heuristiken: Bad-Character-Heuristik und Good-Suffix-Heuristik (5)<br>
•	Boyer-Moore kann große Sprünge im Text machen und Teile des Textes vollständig überspringen<br>
•	Boyer-Moore erreicht eine sub-lineare Laufzeit im Durchschnittsfall<br>
•	Empirische Tests zeigen, dass Boyer-Moore etwa 79,05% effizientere Laufzeiten als Brute-Force aufweist (3)<br>
Die Hauptunterschiede liegen also in der Fähigkeit der fortschrittlicheren Algorithmen, aus bereits gewonnenen Informationen zu lernen und größere Sprünge im Text zu machen, anstatt stur jede Position einzeln zu überprüfen (5,2).<br>

<b><br>5. Anwendungsfälle und Praxisrelevanz<br></b>
Trotz seiner vergleichsweisen geringeren Effizienz hat der Brute-Force-Algorithmus in bestimmten Szenarien durchaus seine Berechtigung:<br>
<b><br>Geeignete Anwendungsfälle:<br></b>
<br>•	Kleine Textmengen: Bei kurzen Texten oder Mustern ist der Overhead der komplexeren Algorithmen möglicherweise größer als der Geschwindigkeitsvorteil (7)<br>
•	Einfache Implementierung: Wenn Einfachheit und schnelle Implementierung wichtiger sind als maximale Effizienz, ist Brute-Force die natürliche Wahl (6)<br>
•	Beschränkter Speicher: Aufgrund seiner minimalen Speicheranforderungen (O(1)) ist Brute-Force für speicherbeschränkte Umgebungen vorteilhaft (9)<br>
•	Einmalsuche: Bei einmaligen Suchen in nicht zu großen Texten ist der Aufwand für Preprocessing möglicherweise nicht gerechtfertigt<br>
•	Lehrkontext: Zur Veranschaulichung grundlegender algorithmischer Konzepte und als Basis für verbesserte Algorithmen (7)<br>
<b><br>Praktische Beispiele:<br></b>
•	Texteditoren für einfache Suchfunktionen<br>
•	Datenvalidierung bei kurzen Eingaben<br>
•	Verarbeitung von Kommandozeilenargumenten<br>
•	Einfache Dokumentanalyse für kleinere Dokumente<br>
•	Als Fallback-Lösung, wenn spezialisierte Algorithmen versagen<br>
<br>In der Praxis werden jedoch für größere Textmengen oder wiederholte Suchen in denselben Texten meist die optimierten Algorithmen KMP oder Boyer-Moore bevorzugt, da sie deutlich bessere Laufzeiten bieten können (3,5).

<b><br>6. Implementierung in Python<br></b>
<br>Im Folgenden wird eine Implementierung des Brute-Force-Algorithmus zur Mustererkennung in Python vorgestellt:<br>



```python
def string_search_bf(*, text, pattern):
    '''Find the first index of pattern in text
    using brute force. If nothing is found, then return None.'''
    n = len(text)
    m = len(pattern)
    for i in range(n-m+1):
        match = True
        for j in range(m):
            if text[i+j] != pattern[j]:
                match = False
                break
        if match:
            return i
string_search_bf(text='hello world', pattern='o')
```

Dieser Code implementiert den Brute-Force-Algorithmus, wie er in der theoretischen Beschreibung erläutert wurde (8). Die äußere Schleife durchläuft alle möglichen Startpositionen im Text, während die innere Schleife den Vergleich mit dem Muster durchführt.<br>

# Knuth Morris Pratt (KMP)

<b><br>1. Definition<br></b>
<br>Der Knuth-Morris-Pratt-Algorithmus (KMP) ist ein effizientes Verfahren zur Mustersuche in Texten, das durch seine lineare Zeitkomplexität von O(n+m) überzeugt – wobei n die Textlänge und m die Länge des Suchmusters (Pattern) ist. Der Algorithmus wurde nach seinen drei Entwicklern, Donald Knuth, James H. Morris und Vaughan Pratt benannt, die ihn 1977 erstmals veröffentlichten (1).<br>
<b><br>2.	FUNKTIONSWEISE<br></b>
<br>Der KMP-Algorithmus beginnt mit einer Vorverarbeitung des Suchmusters (Pattern), bei der eine sogenannte LPS-Tabelle (Longest Prefix Suffix) erstellt wird. Diese Tabelle speichert Informationen über Teilübereinstimmungen innerhalb des Patterns, die es dem Algorithmus ermöglichen, redundante Vergleiche während der eigentlichen Suche zu vermeiden. Durch diese Optimierung kann der Suchvorgang effizienter ablaufen (2).<br>
<b><br>Die Erstellung der LPS-Tabelle läuft wie folgendes:<br></b>
<br>-	Der Suchmuster wird durchgelaufen und für jeden neuen Character wird einen Wert zugewiesen.<br>

-	Wenn der Character zum ersten Mal auftritt, wird dem den Wert 0 zugewiesen.<br>

-	Wenn der Character bereits vorkam, wird eine Wiederholung überprüft<br>

<br>o	Bei einer Wiederholung eines einzelnes Characters wird dem den Wert 1 zugewiesen.<br>

o	Bei einer Wiederholung eines Blockes wird die Länge des Blockes als Wert zugewiesen (z. B. Länge 2 bei sich 2 wiederholende Character)<br>

-	Die Tabelle wird zeilenweise ausgefüllt, bis das gesamte Pattern verarbeitet ist (3).<br> 

<br>Während der Suche nach der Erstellung der LPS-Tabelle vergleicht der KMP-Algorithmus jedes Zeichen in der Textzeichenfolge mit dem entsprechenden Zeichen in der Musterzeichenfolge. Wird eine Nichtübereinstimmung gefunden, verwendet der Algorithmus die Informationen aus der Tabelle der Teilübereinstimmungen, um zu bestimmen, wo der nächste Vergleich beginnen soll (2).<br>
<br>D. h. bei der Suche werden zwei Zeiger initialisiert (z. B. i=0 für den Text, j=0 für das Suchmuster). Der Algorithmus vergleicht die Zeichen an Positionen text[i] und pattern[j]:<br>
<br>	Bei Übereinstimmung:<br>
<br>o	Beide Zeiger werden um 1 erhöht <br>

<br>	Bei Nichtübereinstimmung:<br>
<br>o	Falls j ungleich 0 ist, wird j auf LPS[j-1] gesetzt<br>
<br>o	Falls j = 0, wird nur i um 1 erhöht<br>
	<br>Dieser Prozess wiederholt sich, bis entweder das gesamte Muster gefunden wurde oder 	das Textende erreicht ist (3).<br>
<b><br>Warum bei einer Nichtübereinstimmung wird j auf LPS [j-1] gesetzt? (5)<br></b>
<br>Der Hauptkonzept der LPS-Tabelle besteht darin, die Länge des längsten echten Präfixes zu speichern, das gleichzeitig ein Suffix ist. Eine Nichtübereinstimmung im Suchmuster bedeutet, dass es bis zur Position j-1 eine Übereinstimmung gab. Um einen vollständigen Neustart der Suche zu vermeiden, gibt der LPS[j-1] -Wert an, wie viele Zeichen des Musters bereits als Präfix-Suffix-Kombination bekannt sind. Der Algorithmus springt zur Position LPS[j-1] im Muster und vergleicht ab dort weiter, ohne bereits geprüfte Zeichen erneut zu betrachten.<br>

<b><br>3.	ANALYSE DER ZEITKOMPLEXITÄT<br></b>
<br>Für die Analyse der Zeitkomplexität des KMP-Algorithmus wird eine Variable k initialisiert, die die Differenz zwischen i und j speichert. Während der Suche gilt:<br>
<br>	Bei einer Übereinstimmung werden beide Zeiger i und j um eins erhöht, sodass k unverändert bleibt.<br>
<br>	Bei einer Nichtübereinstimmung und j > 0 steigt k mindestens um eins und wird auf Differenz zwischen i und LPS[j-1] gesetzt.<br>
<br>	Bei einer Nichtübereinstimmung und j = 0 erhöht sich i ums eins, wodurch auch k um eins steigt.
Die maximale Anzahl der Durchgänge beträgt höchstens 2n, sodass die gesamte Zeitkomplexität des KMP-Algorithmus O(n+m) beträgt (3).<br>
	

<b><br>4.	UNTERSCHIED ZU ANDEREN PATTERN-MATCHING-ALGORITHMEN<br></b>

# Boyer-Moore
<b><br>Beschreibung des Boyer-Moore-Algorithmus<br></b>
<br>Der Boyer-Moore-Algorithmus ist ein effizienter Suchalgorithmus, der verwendet wird, um ein Muster in einem längeren Text zu finden. Im Gegensatz zu einfachen Algorithmen wie Brute-Force vergleicht er das Muster von rechts nach links und nutzt dabei zwei Heuristiken, um große Teile des Textes zu überspringen:<br>
<br>•	Bad-Character-Heuristik: Wenn ein Zeichen nicht passt, wird das Muster basierend auf diesem Zeichen verschoben.<br>
<br>•	Good-Suffix-Heuristik: Wenn ein Teil des Musters bereits gepasst hat, wird das Muster zur nächsten passenden Stelle verschoben.<br>
<br>Dadurch werden unnötige Vergleiche vermieden und der Algorithmus ist in der Praxis oft sehr schnell.<br>
<b><br>Effizienz:<br></b>
<br>•	Best Case: O(n/m)<br>
•	Worst Case: O(n × m), aber selten<br>
•	Praktisch: sehr schnell, besonders bei großem Text<br>
<b><br>Vorteile:<br></b>
<br>•	Sehr effizient bei langen Texten<br>
•	Überspringt viele Zeichen<br>
<b><br>Nachteile:<br></b>
<br>•	Komplexer zu implementieren<br>
•	Nicht optimal bei kurzen Mustern<br>
<br>
<b><br>Main Code<br></b>

```python
import time
# Funktion zum Messen der Ausführungszeit
def measure_time(func, text, pattern):
    start = time.time()
    result = func(text, pattern)
    end = time.time()
    return result, end - start
# Boyer-Moore-Algorithmus mit Bad Character Heuristik
def boyer_moore(text, pattern):
    m = len(pattern)
    n = len(text)
    if m == 0:
        return 0
    # Schritt 1: Bad Character Heuristik vorbereiten
    bad_char = [-1] * 256
    for i in range(m):
        bad_char[ord(pattern[i])] = i
    # Schritt 2: Suche im Text
    s = 0  # Verschiebung des Musters im Text
    while s <= n - m:
        j = m - 1
        # Vergleiche von rechts nach links
        while j >= 0 and pattern[j] == text[s + j]:
            j -= 1
        if j < 0:
            return s  # Muster gefunden
        else:
            skip = max(1, j - bad_char[ord(text[s + j])])
            s += skip
    return -1  # Muster nicht gefunden
```

<b><br>8.1. Vergleich der betrachteten Algorithmen<br></b>
<br>Der Brute-Force-Algorithmus ist der einfachste und intuitivste Ansatz, bei dem jedes mögliche Teilfenster des Textes zeichenweise mit dem Muster verglichen wird. Seine Stärken liegen in der universellen Anwendbarkeit, dem minimalen Speicherbedarf und der einfachen Implementierung. Diese Eigenschaften machen ihn besonders geeignet für didaktische Zwecke oder als Baseline-Referenz. Allerdings zeigt sich seine Schwäche in der ineffizienten Zeitkomplexität von O(n·m) im Worst-Case, die bei großen Texten oder Mustern zu erheblichen Leistungseinbußen führt. Dies wurde im Experiment deutlich, als bei einem repetitiven Text überflüssige Vergleiche zu einer längeren Ausführungszeit führten. Dennoch bleibt der Brute-Force-Algorithmus relevant für kleine Datenmengen oder einfache Anwendungsfälle.<br>
<br>Der Knuth-Morris-Pratt-Algorithmus (KMP) verbessert die Effizienz durch ein Präfix-Array (LPS), das Wiederholungen im Muster erkennt und redundante Vergleiche vermeidet. Dadurch wird die Zeitkomplexität auf O(n + m) reduziert, was insbesondere bei Daten mit hoher Redundanz wie DNA-Sequenzen von Vorteil ist. Der Algorithmus benötigt jedoch zusätzlichen Speicher für das LPS-Array und ist weniger effektiv bei großen Alphabeten oder kurzen Mustern. Die experimentellen Ergebnisse zeigen, dass KMP besonders effizient ist, wenn das Muster viele Wiederholungen enthält, da er den Text ohne Backtracking durchläuft.<br>
<br>Der Boyer-Moore-Algorithmus nutzt zwei Heuristiken – die Bad-Character-Regel und die Good-Suffix-Regel –, um das Muster rückwärts zu vergleichen und große Sprünge im Text zu ermöglichen. Diese Strategie führt zu einer durchschnittlichen Zeitkomplexität von O(n/m), was ihn besonders geeignet für natürlichsprachliche Texte macht. Der Algorithmus ist ideal für lange Muster und große Alphabete, da er bei Mismatches mehrere Zeichen überspringen kann. Allerdings ist das Preprocessing komplexer als bei KMP und erfordert mehr Speicherplatz für die Heuristik-Tabellen. Zudem kann es in seltenenmn Fällen zu einer ineffizienten Worst-Case-Laufzeit von O(n·m) kommen.<br>

<b><br>8.2 Interpretation der experimentellen Ergebnisse<br></b>
<br>Die experimentellen Ergebnisse bestätigen die theoretischen Analysen: Während Brute-Force bei kleinen Datensätzen praktikabel bleibt, zeigt KMP seine Stärken in Szenarien mit hoher Redundanz wie DNA-Sequenzen. Boyer-Moore hingegen glänzt in natürlichsprachlichen Texten durch seine praktische Effizienz und Skalierbarkeit.<br>
<br>Die Wahl des optimalen Algorithmus hängt stark vom Anwendungsfall ab. Faktoren wie die Größe des Textes und des Musters, die Struktur der Daten sowie Anforderungen an Geschwindigkeit und Speicherverbrauch spielen eine entscheidende Rolle. Während der Brute-Force-Algorithmus aufgrund seiner Einfachheit weiterhin relevante Anwendungsfälle besitzt, dominieren KMP und Boyer-Moore in professionellen Umgebungen aufgrund ihrer Effizienz.<br>
<b><br>8.3 Ausblick auf zukünftige Entwicklungen<br></b>
<br>Zukünftige Entwicklungen könnten hybride Ansätze umfassen, die die Vorteile verschiedener Algorithmen kombinieren – beispielsweise durch die Nutzung des LPS-Arrays von KMP zur Vermeidung redundanter Vergleiche und der Bad-Character-Heuristik von Boyer-Moore zur Beschleunigung der Suche. Parallelisierung auf modernen Hardwareplattformen wie GPUs oder FPGA-basierte Implementierungen könnte Pattern-Matching-Aufgaben in Big-Data-Szenarien weiter optimieren. Adaptive Algorithmen könnten mithilfe von maschinellem Lernen automatisch den effizientesten Algorithmus basierend auf den Eingabedaten auswählen. Darüber hinaus könnten domänenspezifische Optimierungen entwickelt werden, etwa für die Genomsequenzierung oder Intrusion-Detection-Systeme in der Netzwerksicherheit.<br>
<b><br>8.4 Ethische Implikationen<br></b>
<br>Die Weiterentwicklung von Pattern-Matching-Algorithmen wirft auch ethische Fragen auf, insbesondere im Hinblick auf Datenschutz und Privatsphäre bei Anwendungen wie Massenüberwachung oder Genomdatenanalyse. Zukünftige Arbeiten sollten daher nicht nur die Effizienz dieser Algorithmen verbessern, sondern auch sicherstellen, dass ethische Richtlinien eingehalten werden.<br>

<b><br>8.6 Abschließende Bewertung<br></b>

<br>Zusammenfassend lässt sich sagen, dass Pattern-Matching-Algorithmen eine unverzichtbare Grundlage moderner Datenverarbeitungssysteme darstellen. Während Brute-Force als konzeptionelle Basis weiterhin relevant bleibt, bieten KMP und Boyer-Moore leistungsstarke Alternativen für anspruchsvolle Anwendungen. Die experimentellen Ergebnisse unterstreichen jedoch, dass kein Algorithmus universell optimal ist – vielmehr bestimmen die spezifischen Anforderungen eines Use-Cases die Wahl der Methode. Zukünftige Forschung wird entscheidend dazu beitragen, diese Algorithmen an die wachsenden Herausforderungen der Datenintensität und Echtzeitanforderungen anzupassen.<br>
